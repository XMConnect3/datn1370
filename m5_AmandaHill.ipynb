```python
import pandas as pd               # tables
import seaborn as sns             # charts
import matplotlib.pyplot as plt   # titles and labels
```

```python
df = pd.read_csv(
    "https://raw.githubusercontent.com/papayuh/datn1370-datasets/refs/heads/main/data/synthetic/coffee_shop_sales.csv")
df.head()
```

```python
df.groupby("shop") ["revenue_usd"].describe().round(1)
```

```python
sns.barplot(data=df, x="shop", y="revenue_usd", errorbar=None)
plt.title("On average, the Campus shop has the lowest revenue")
plt.xlabel("SHOPS")
plt.ylabel("Coffee Shop Average Daily Revenue")
plt.show()
```

The Campus shop has the lowest daily average revenue with $845, followed by Riverside with $1,100, and leading with the higest is Downtown with $1190. I chose the bar chart because it is easier to compare the coffee shops daily average at a glance.

```python
ax1 = sns.regplot(data=df, x="temp_high_f", y="iced_drinks")
plt.title("Iced Drink Sales Tend to Increase as the Temperature Rises")
plt.xlabel("Temperature Daily Highest F")
plt.ylabel("Iced Drinks Sold")
plt.show()
```

```python
corr = df[["temp_high_f", "iced_drinks"]].corr()
display(corr)
```

The scatterplot slopes up, and shows a correlation between the temperature and the number of iced drinks sold. Because r is only 0.69 and the chart combines all 3 coffee shops, I believe other factors are also driving the iced drink sales.  I used the scatterplot chart because I wanted to determine if temperature and iced drink sales were correlated across the shops and this provided a vivid picture.
